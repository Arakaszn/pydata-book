### Question 2: Styling and Iterative Plotting
**Objective:** Practice using colors, markers, line styles, and the concept of drawing multiple series on the same axes.

**The Task:**
1.  Create a single plot (one subplot).
2.  Generate a random walk of 30 steps using `numpy.random.standard_normal(30).cumsum()`.
3.  Plot this data with the following specific styling:
    * **Color:** Green
    * **Line Style:** Dashed (`--`)
    * **Marker:** Circle (`o`)
4.  Add a second line to the *same plot* using a different random walk, but this time use the **draw style** `'steps-post'` to show the difference between standard interpolation and step plots.


```python
# Type your solution here:

import matplotlib.pyplot as plt
import numpy as np

walk1 = np.random.standard_normal(30).cumsum()
walk2 = np.random.standard_normal(30).cumsum()
fig, ax = plt.subplots(figsize=(10, 6))

# First walk plot
ax.plot(walk1, color='green', linestyle='--', marker='o', label='Walk 1 (Standard Interpolation)')

# second walk plot
ax.plot(walk2, color='blue', drawstyle='steps-post', label='Walk 2 (Steps-post)')

# Tittle and labels
ax.set_title('Styling and Iterative Plotting: Random Walks')
ax.set_xlabel('Steps')
ax.set_ylabel('Cumulative Sum')
ax.legend()
ax.grid(True, alpha=0.3)

plt.savefig('random_walks.png', bbox_inches='tight')
```

### Question 1: Plotting Series with Pandas
**Objective:** Master the `s.plot()` method and understand how pandas handles index-based labeling automatically.

**The Task:**
1.  Create a pandas `Series` containing a random walk of 100 values (use `np.random.standard_normal(100).cumsum()`).
2.  Set the index of this series to be a date range starting from `'2026-01-01'` for 100 periods.
3.  Use the pandas `.plot()` method to create a line plot.
4.  **Requirement:** Enable the grid and set the title to *"100 Days of Random Flux"*.

**Key takeaway:** Notice how pandas automatically uses the Series index for the x-axis labels and the Series name (if provided) for the legend.

```python
# Type your solution here:

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = np.random.standard_normal(100).cumsum()

dates = pd.date_range(start='2026-01-01', periods=100)

series = pd.Series(data, index=dates, name="Flux")

plt.figure(figsize=(10, 6))
series.plot(grid=True, title="100 Days of Random Flux")

plt.ylabel("Value")
plt.xlabel("Date")

plt.savefig('pandas_random_flux.png', bbox_inches='tight')
```

### Question 2: Bar Plots and Grouped Data
**Objective:** Understand how DataFrames are rendered as grouped bar charts and how to manipulate the orientation.

**The Task:**
1.  Create a DataFrame with 5 rows and 3 columns using `np.random.uniform(0, 1, (5, 3))`. 
2.  Name the columns `['A', 'B', 'C']` and the index `['V1', 'V2', 'V3', 'V4', 'V5']`.
3.  Generate two separate plots:
    * **Plot A:** A standard vertical grouped bar plot.
    * **Plot B:** A horizontal bar plot where the bars for A, B, and C are **stacked** on top of each other.
4.  Apply a specific alpha (transparency) of `0.7` to both plots.


```python
# Type your solution here:

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(0, 1, (5, 3))
df = pd.DataFrame(data, columns=['A', 'B', 'C'], index=['V1', 'V2', 'V3', 'V4', 'V5'])

plt.figure(figsize=(8, 5))
df.plot(kind='bar', alpha=0.7, ax=plt.gca())
plt.title("Plot A: Vertical Grouped Bar Plot")
plt.ylabel("Values")
plt.xlabel("Categories")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('plot_a.png')

# plot b
plt.figure(figsize=(8, 5))
df.plot(kind='barh', stacked=True, alpha=0.7, ax=plt.gca())
plt.title("Plot B: Horizontal Stacked Bar Plot")
plt.xlabel("Values")
plt.ylabel("Categories")
plt.tight_layout()
plt.savefig('plot_b.png')
```
