Given the following DataFrame representing a small sales team:
```python

import numpy as np

df = pd.DataFrame({
    'Region': ['East', 'West', 'East', 'South', 'West', 'South'],
    'Sales': [100, 250, 150, 300, 200, 50],
    'Expenses': [80, 200, 120, 250, 150, 40]
})
```
1. Write a single line of code to calculate the **mean** Sales for each **Region**.
2. Explain in your own words what the "Split" and "Apply" steps are doing specifically for the 'East' region in this calculation.


````python
# Type your solution here:

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Region': ['East', 'West', 'East', 'South', 'West', 'South'],
    'Sales': [100, 250, 150, 300, 200, 50],
    'Expenses': [80, 200, 120, 250, 150, 40]
})

regional_mean = df.groupby('Region').mean()
regional_mean

# Split and apply are two steps in the groupby process. They are doing the following for the East region: They are splitting the data into groups based on the 'Region' column, and then applying the mean function to calculate the average Sales and Expenses for each region. The result is a new DataFrame that shows the average Sales and Expenses for each region.
````

## Question 1: Group-Level Top-N Analysis

Suppose you have a DataFrame of sales data with columns `['Region', 'Salesperson', 'Revenue']`. Write a function called `top_sales` that takes a group and returns the rows containing the **top 2** highest `Revenue` values. Then, group the DataFrame by `Region` and use `.apply()` to see the top performers in every region.

**Data Setup for Students:**
```python
import pandas as pd

data = pd.DataFrame({
    'Region': ['East', 'East', 'East', 'West', 'West', 'West', 'South', 'South'],
    'Salesperson': ['Alice', 'Bob', 'Charlie', 'David', 'Edward', 'Frank', 'Guy', 'Helen'],
    'Revenue': [500, 700, 600, 300, 800, 450, 900, 1000]
})
```

````python
# Type your solution here:

import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Region': ['East', 'East', 'East', 'West', 'West', 'West', 'South', 'South'],
    'Salesperson': ['Alice', 'Bob', 'Charlie', 'David', 'Edward', 'Frank', 'Guy', 'Helen'],
    'Revenue': [500, 700, 600, 300, 800, 450, 900, 1000]
})

# Define the custom function
def top_sales(group):
    return group.nlargest(2, 'Revenue')

top_performers = data.groupby('Region', group_keys=False).apply(top_sales)

top_performers
````

## Question 2: Group-Specific Normalization
**Objective:** Understand how to perform "Broadcast" operations within groups (Standardization).

Using the same sales data, you need to compare performance relative to the group average. Write a function that **normalizes** the `Revenue` within each group using the following formula:
$$Revenue_{norm} = \frac{Revenue - \text{mean}}{\text{std}}$$
Apply this function to each `Region` so that students can see which salesperson outperformed their specific regional average.

````python
# Type your solution here:

import pandas as pd

data = pd.DataFrame({
    'Region': ['East', 'East', 'East', 'West', 'West', 'West', 'South', 'South'],
    'Salesperson': ['Alice', 'Bob', 'Charlie', 'David', 'Edward', 'Frank', 'Guy', 'Helen'],
    'Revenue': [500, 700, 600, 300, 800, 450, 900, 1000]
})

def normalize_revenue(group):
    """Normalizes the Revenue column within a specific group."""
    avg = group['Revenue'].mean()
    std_dev = group['Revenue'].std()
    
    group['Normalized_Revenue'] = (group['Revenue'] - avg) / std_dev
    return group

#Group the data by 'Region' and apply the normalization function to each group
normalized_data = data.groupby('Region', group_keys=False).apply(normalize_revenue)
normalized_data
````

### Question 1: The `transform` Method

You have a DataFrame of student grades from different departments. Create a new column called `deviance` that calculates how much each student's score differs from their **department's average score**.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'dept': ['CS', 'CS', 'Math', 'Math', 'CS'],
    'score': [85, 95, 70, 80, 75]
})

# Your code here:
# 1. Group by 'dept'
# 2. Use .transform() to subtract the group mean from the 'score' column
```

````python
# Type your solution here:

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'dept': ['CS', 'CS', 'Math', 'Math', 'CS'],
    'score': [85, 95, 70, 80, 75]
})

# Calculate the mean for each department and subtract it from the original score
df['deviance'] = df['score'] - df.groupby('dept')['score'].transform('mean')
df
````

## Exercise 2: Group-wise Function Application with `apply`

Using the same `data` DataFrame from Exercise 1, write a function that selects the **top N** highest-paid employees in each department. Then, use the `groupby` and `apply` method to return a DataFrame containing the top 1 salary for every department.

```python
def top_salary(df, n=1):
    return df.sort_values(by='Salary', ascending=False)[:n]

# Your code here:
```

````python
# Type your solution here:

import pandas as pd

data = pd.DataFrame({
    'Dept': ['HR', 'HR', 'IT', 'IT', 'IT', 'Sales'],
    'Salary': [50000, 60000, 70000, 80000, 90000, 55000]
})

# Define the function to sort and slice the top 'n' rows
def top_salary(df, n=1):
    return df.sort_values(by='Salary', ascending=False)[:n]

# Group by 'Dept' and apply the function, passing n=1 as an argument
top_earners = data.groupby('Dept', group_keys=False).apply(top_salary, n=1)
top_earners
````
